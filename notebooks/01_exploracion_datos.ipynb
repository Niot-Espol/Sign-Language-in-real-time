Consolida las capturas de `data/raw` y revisa el balance de clases.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RAIZ = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
import sys
sys.path.insert(0, str(RAIZ))
from src.data_collection.consolidar_dataset import consolidar

salida = RAIZ / 'data' / 'processed' / 'dataset_consolidado.csv'
dataset, metadata = consolidar(RAIZ / 'data' / 'raw', salida)
print(f'{len(dataset)} muestras guardadas en {salida}')
display(dataset.head())
```

```python
conteo = dataset['label'].value_counts().sort_index()
display(conteo.to_frame('muestras'))
conteo.plot.bar(title='Muestras por seña', ylabel='Cantidad', figsize=(10, 4))
plt.tight_layout()
plt.show()
```
