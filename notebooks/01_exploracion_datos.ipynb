# 01 - Exploración de datos
Consolida las capturas de `data/raw` y revisa el balance de clases.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RAIZ = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
archivos = sorted((RAIZ / 'data' / 'raw').glob('*.csv'))
if not archivos:
    raise FileNotFoundError('No hay capturas CSV en data/raw')

dataset = pd.concat((pd.read_csv(archivo) for archivo in archivos), ignore_index=True)
dataset = dataset.dropna().drop_duplicates().reset_index(drop=True)
salida = RAIZ / 'data' / 'processed' / 'dataset.csv'
salida.parent.mkdir(parents=True, exist_ok=True)
dataset.to_csv(salida, index=False)
print(f'{len(dataset)} muestras guardadas en {salida}')
display(dataset.head())
```

```python
conteo = dataset['label'].value_counts().sort_index()
display(conteo.to_frame('muestras'))
conteo.plot.bar(title='Muestras por seña', ylabel='Cantidad', figsize=(10, 4))
plt.tight_layout()
plt.show()
```
